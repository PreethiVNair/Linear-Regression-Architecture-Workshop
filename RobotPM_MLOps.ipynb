### Explanation

- `src/` contains modular pipeline components  
- `configs/` controls all parameters  
- `experiments/` stores reproducible experiment outputs  
- `data/` stores raw + processed datasets  
- `RobotPM_MLOps.ipynb` documents architecture  
- `LinearRegression.ipynb` contains actual ML implementation  


## Architectural Goals

### Separation of Concerns
Each module performs one job only.

### Config-Driven Execution
All parameters come from YAML files.

### Experiment Tracking
Every run produces a reproducible record.

### Reproducibility
Pinned versions, fixed seeds, deterministic pipeline.
